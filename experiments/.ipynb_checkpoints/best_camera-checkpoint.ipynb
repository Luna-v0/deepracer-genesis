# Best Camera Model — HPO + Final Training + Evaluation

One self-contained experiment: find hyperparameters that make **end-to-end
vision** (camera → CNN → steering/throttle) actually learn, then train the
best configuration at scale with the full sim2real DR stack, evaluate it,
and export an ONNX bundle for the physical car.

**Why HPO is the core here, not a garnish:** every camera run recorded in
this repo so far ended at **0% lap completion** — including a full 10M-step
`CameraMadronaDr` run (0.6 m mean progress) and the whole
`experiments/camera_std_sweep.py` diagnostic family (~1.9 m, off-track rate
1.0). Feature-vector policies complete 100% of laps in the very same
environment, so the task and reward are fine — the vision learning setup is
what's broken. The sweep established that the action-std dynamics are only
stable with a `std_range=(0.1, 1.0)` ceiling; this study keeps that fix
fixed and searches the remaining first-order suspects (lr, entropy, frame
stack, discount, batch shape) with **no DR** and a single track, so trials
measure pure learning ability. The winner then gets the full CameraZoo-style
DR treatment for the long run.

**Reproducibility:** a single `SEED` drives python/numpy/torch, the env
build, the Optuna sampler, and the train/test track split. Every GPU stage
runs in its own subprocess (crash isolation + clean VRAM) with the params
recorded as JSON next to its log under `runs/best_camera/logs/`.

**Renderer:** measured, not assumed — the notebook benches Madrona vs Nyx
throughput below and picks Nyx (better photorealism → better sim2real for
the physical car) unless it is too slow to fit the wall-clock budget.


```python
# --- experiment configuration (papermill parameters cell) ------------------
SEED = 0
ROOT = "runs/best_camera"          # everything lands under here
WALL_BUDGET_H = 10.0               # total wall-clock allowance

# throughput bench
BENCH_STEPS = 150                  # timed control steps per config
BENCH_CONFIGS = [("madrona", 64), ("madrona", 128), ("nyx", 64), ("nyx", 128)]
NYX_SLOWDOWN_OK = 2.0              # prefer nyx unless > this x slower than madrona

# HPO stage (no-DR, single track: measures pure learning ability)
HPO_TRIALS = 16
HPO_DEADLINE_H = 3.4               # stop launching trials past this elapsed time
TRIAL_STEPS = 1_200_000
TRIAL_EVAL_EVERY = 600_000
TRIAL_TRACK = "reinvent_base"

# final run (winner config + zoo + full DR)
FINAL_STEPS_MAX = 30_000_000
FINAL_STEPS_MIN = 2_000_000        # never train below a token run
FINAL_RESERVE_H = 1.0              # kept back for analysis + video + export
FINAL_EVALS = 6                    # periodic evals across the final run

RESOLUTION = (160, 120)            # matches the physical car camera pipeline
```

```python
# Parameters
WALL_BUDGET_H = 7.8
```

```python
"""Session plumbing: seeds, paths, and the subprocess launcher."""

import json
import math
import os
import subprocess
import sys
import time

import numpy as np

np.random.seed(SEED)

T0 = time.time()
LOGS = os.path.join(ROOT, "logs")
os.makedirs(LOGS, exist_ok=True)


def elapsed_h() -> float:
    """Wall-clock hours since the notebook started.

    Returns:
        Elapsed time in hours.
    """
    return (time.time() - T0) / 3600.0


def budget_left_h(reserve: float = 0.0) -> float:
    """Remaining wall-clock hours of the experiment budget.

    Args:
        reserve: Hours to keep back (final analysis/export).

    Returns:
        Hours still available for training work (floored at 0).
    """
    return max(0.0, WALL_BUDGET_H - elapsed_h() - reserve)


def launch(name: str, source: str, params: dict, timeout_s: float):
    """Run one GPU stage in an isolated subprocess and return its result.

    The stage gets its parameters as a JSON file path in ``argv[1]``, logs
    stdout+stderr to ``runs/best_camera/logs/<name>.log``, and reports back
    through the JSON file named by ``params["out"]``. Isolation keeps a
    genesis crash (or VRAM fragmentation) contained to the stage.

    Args:
        name: Stage name; names the log/params/result files.
        source: Python source executed via ``python -c``.
        params: JSON-serializable stage parameters.
        timeout_s: Kill the stage after this many seconds.

    Returns:
        The stage's result dict, or ``{"error": ...}`` on crash/timeout.
    """
    params = dict(params, out=os.path.join(LOGS, f"{name}.result.json"))
    ppath = os.path.join(LOGS, f"{name}.params.json")
    with open(ppath, "w") as f:
        json.dump(params, f, indent=2)
    if os.path.exists(params["out"]):
        os.remove(params["out"])
    t0 = time.time()
    with open(os.path.join(LOGS, f"{name}.log"), "w") as log:
        try:
            proc = subprocess.run([sys.executable, "-c", source, ppath],
                                  stdout=log, stderr=subprocess.STDOUT,
                                  cwd=os.getcwd(), timeout=timeout_s)
        except subprocess.TimeoutExpired:
            return {"error": f"timeout after {timeout_s:.0f}s"}
    if not os.path.exists(params["out"]):
        return {"error": f"no result (exit {proc.returncode}) — "
                         f"see {LOGS}/{name}.log"}
    with open(params["out"]) as f:
        result = json.load(f)
    result["wall_s"] = round(time.time() - t0, 1)
    return result


print(f"root={ROOT}  seed={SEED}  budget={WALL_BUDGET_H}h  python={sys.version.split()[0]}")
```

```python
"""Subprocess payloads: one shared spec builder + per-mode bodies.

The notebook process itself never imports genesis — every GPU stage
(frames, bench, trial, final, video) runs the RUNNER payload in its own
process, and the ONNX export runs the EXPORT payload (which must stay
genesis-free: genesis and onnxruntime bundle clashing LLVM symbols).
"""

_PRELUDE = r"""
import json, os, sys
params = json.load(open(sys.argv[1]))
os.environ.setdefault("PYTORCH_CUDA_ALLOC_CONF", "expandable_segments:True")
import numpy as np
import torch
np.random.seed(params["seed"]); torch.manual_seed(params["seed"])
if params.get("compile_zoo"):
    from deepracer_genesis.tools.zoo import compile_zoo, demo_zoo
    compile_zoo(demo_zoo())

from deepracer_genesis.experiment import (
    AsymmetricCameraPolicy, CameraEnvironment, DomainRandomizationActions,
    DomainRandomizationCamera, DomainRandomizationPhysics,
    DomainRandomizationTrackAppearance, Evaluation, PPO, run,
)


def build_spec(p):
    "Assemble the camera ExperimentSpec for one stage from spec params p."
    pipe = CameraEnvironment(render=p["render"], resolution=tuple(p["resolution"]),
                             num_envs=p["num_envs"], tracks=tuple(p["tracks"]),
                             frame_stack=p["frame_stack"],
                             random_direction=p.get("random_direction", False))
    if p.get("dr"):
        pipe = (pipe
                >> DomainRandomizationTrackAppearance(strength=0.6)
                >> DomainRandomizationCamera(brightness=(0.7, 1.3), hue=0.05,
                                             blur=0.3, camera_jitter=True)
                >> DomainRandomizationPhysics())
    pipe = pipe >> AsymmetricCameraPolicy(
        actor_keys=("camera",), critic_keys=("camera", "state"),
        distribution={"std_range": (0.1, 1.0)})
    if p.get("dr"):
        pipe = pipe >> DomainRandomizationActions(
            steer_noise=0.02, speed_noise=0.05, delay_steps=1)
    pipe = pipe >> PPO(lr=p["lr"], entropy_coef=p["entropy_coef"],
                       epochs=p["epochs"], minibatches=p["minibatches"],
                       gamma=p["gamma"])
    if p.get("real_tracks"):
        pipe = pipe >> Evaluation(real_tracks=tuple(p["real_tracks"]),
                                  charts=True)
    return pipe.build(seed=params["seed"], total_env_steps=p["steps"],
                      eval_every_steps=p["eval_every"],
                      ablation_group=p["group"], variant=p["variant"])


def _finish(result):
    "Write the stage result JSON."
    with open(params["out"], "w") as f:
        json.dump(result, f)
"""

_RUNNER_BODY = r"""

class _ConstActor:
    "Constant mild-throttle actor for warmup/bench rollouts."

    def __init__(self, sim):
        self.act = torch.zeros(sim.num_envs, 2, device=sim.device)
        self.act[:, 1] = 0.4

    def __call__(self, td):
        td.set("action", self.act)
        return td


mode = params["mode"]
try:
    if mode in ("frames", "bench"):
        from deepracer_genesis.experiment.builder import Builder
        from deepracer_genesis.experiment.evaluator import evaluate_policy
        spec = build_spec(params["spec"])
        sim = Builder(spec).sim()
        actor = _ConstActor(sim)
        with torch.inference_mode():
            evaluate_policy(sim, actor, steps=30)          # warmup / JIT
            if mode == "frames":
                td = sim.get_observations()
                cam = td["camera"][:params["n_frames"]].float().cpu().numpy()
                np.save(params["out_npy"], cam)
                _finish({"shape": list(cam.shape),
                         "min": float(cam.min()), "max": float(cam.max())})
            else:
                import time as _t
                t0 = _t.perf_counter()
                evaluate_policy(sim, actor, steps=params["bench_steps"])
                dt = _t.perf_counter() - t0
                _finish({"env_steps_per_s": round(
                    params["bench_steps"] * sim.num_envs / dt, 1)})
    elif mode in ("trial", "final"):
        spec = build_spec(params["spec"])
        record = run(spec, root=params["root"])
        _finish({"metrics": record.metrics,
                 "history": record.eval_history,
                 "holdout": record.holdout,
                 "train": record.train,
                 "run_dir": spec.run_dir(params["root"])})
    elif mode == "video":
        from deepracer_genesis.experiment.visualize import rollout_video
        spec = build_spec(params["spec"])
        path = rollout_video(spec, root=params["root"],
                             ckpt=params["ckpt"], steps=params["video_steps"],
                             num_envs=params.get("video_envs", 8))
        _finish({"video": path})
    else:
        _finish({"error": f"unknown mode {mode}"})
except Exception as e:                     # noqa: BLE001 - reported, not raised
    import traceback
    traceback.print_exc()
    _finish({"error": f"{type(e).__name__}: {e}"})
"""

_EXPORT_BODY = r"""

try:
    assert "genesis" not in sys.modules, "export payload must stay genesis-free"
    from deepracer_genesis.deploy.onnx import export_policy
    spec = build_spec(params["spec"])
    out = export_policy(spec, root=params["root"], ckpt=params["ckpt"],
                        bundle_name=params["bundle_name"])
    _finish({"export_dir": out, "files": sorted(os.listdir(out))})
except Exception as e:                     # noqa: BLE001 - reported, not raised
    import traceback
    traceback.print_exc()
    _finish({"error": f"{type(e).__name__}: {e}"})
"""

RUNNER_SRC = _PRELUDE + _RUNNER_BODY
EXPORT_SRC = _PRELUDE + _EXPORT_BODY


def base_spec_params(**overrides) -> dict:
    """Baseline spec params every stage starts from.

    Args:
        **overrides: Spec fields to override (render, num_envs, ppo, ...).

    Returns:
        The merged spec params dict for a payload's ``build_spec``.
    """
    p = {
        "render": "madrona", "resolution": list(RESOLUTION), "num_envs": 64,
        "tracks": [TRIAL_TRACK], "frame_stack": 4, "dr": False,
        "lr": 3.0e-4, "entropy_coef": 3.0e-3, "epochs": 5, "minibatches": 4,
        "gamma": 0.99, "steps": TRIAL_STEPS, "eval_every": TRIAL_EVAL_EVERY,
        "group": "best_camera", "variant": "probe",
    }
    p.update(overrides)
    return p
```

```python
"""Sanity: what the car actually sees (before spending hours training on it)."""

import matplotlib.pyplot as plt

r = launch("frames", RUNNER_SRC, {
    "mode": "frames", "seed": SEED, "n_frames": 8,
    "out_npy": os.path.join(LOGS, "frames.npy"),
    "spec": base_spec_params(num_envs=8),
}, timeout_s=600)
print(r)
assert "error" not in r, f"frames stage failed: {r}"

cam = np.load(os.path.join(LOGS, "frames.npy"))
rgb = np.transpose(cam[:, -3:], (0, 2, 3, 1))       # newest frame of the stack
rgb = (rgb - rgb.min()) / (rgb.max() - rgb.min() + 1e-9)
fig, axes = plt.subplots(2, 4, figsize=(14, 5.5))
for i, ax in enumerate(axes.flat):
    ax.imshow(rgb[i]); ax.set_title(f"env {i}", fontsize=9); ax.axis("off")
fig.suptitle("onboard camera — newest stacked frame per env")
fig.tight_layout(); plt.show()
```

```python
"""Renderer bench: measure Madrona vs Nyx, then decide with numbers."""

bench = {}
for render, n in BENCH_CONFIGS:
    r = launch(f"bench_{render}_{n}", RUNNER_SRC, {
        "mode": "bench", "seed": SEED, "bench_steps": BENCH_STEPS,
        "spec": base_spec_params(render=render, num_envs=n),
    }, timeout_s=900)
    bench[(render, n)] = r.get("env_steps_per_s", None)
    print(f"{render:8s} x{n:4d}: {bench[(render, n)]} env-steps/s"
          + (f"   ({r['error']})" if "error" in r else ""))


def pick_renderer(bench: dict) -> tuple:
    """Choose (render, num_envs) from the measured throughput table.

    Nyx wins (better photorealism → better sim2real) unless it is more than
    ``NYX_SLOWDOWN_OK``x slower than the best Madrona config or broken.

    Args:
        bench: ``{(render, num_envs): env_steps_per_s | None}``.

    Returns:
        The chosen ``(render, num_envs)`` pair.
    """
    best = {}
    for (render, n), sps in bench.items():
        if sps and (render not in best or sps > best[render][0]):
            best[render] = (sps, n)
    assert best, "no renderer produced a benchmark result"
    if "nyx" in best and "madrona" in best:
        if best["madrona"][0] / best["nyx"][0] <= NYX_SLOWDOWN_OK:
            return ("nyx", best["nyx"][1])
        print(f"nyx is {best['madrona'][0]/best['nyx'][0]:.1f}x slower than "
              f"madrona (> {NYX_SLOWDOWN_OK}x budget rule) — using madrona")
        return ("madrona", best["madrona"][1])
    render = next(iter(best))
    return (render, best[render][1])


RENDER, NUM_ENVS = pick_renderer(bench)
SPS_EST = bench[(RENDER, NUM_ENVS)]
print(f"\ndecision: render={RENDER} num_envs={NUM_ENVS} (~{SPS_EST:.0f} env-steps/s)")
```

## HPO — make vision learn at all

Short no-DR trials on a single track, so the objective isolates *learning
ability*. The metric is **`mean_progress_m`** at the trial's final
deterministic eval — completion is 0 for nearly all camera configs so far,
while progress is a dense, discriminative signal (the historical plateau to
beat is ~1.9 m).

Fixed (evidence-based): `std_range=(0.1, 1.0)` (the only stable std dynamics
found by `camera_std_sweep`), resolution 160×120 (physical-camera parity),
renderer + `num_envs` from the bench above. Searched: `lr`,
`entropy_coef`, `epochs`, `minibatches`, `frame_stack` (1 vs 4 — the
velocity-blindness hypothesis), `gamma`. The first trial is pinned to the
sweep's best-known configuration as an anchor; the TPE sampler is seeded.
Each trial is a subprocess: a crash marks the trial failed and the study
continues.


```python
"""The Optuna study (resumable: sqlite storage under the run root)."""

import optuna

optuna.logging.set_verbosity(optuna.logging.WARNING)

# NOTE: the bench measures ROLLOUT-only throughput; PPO training runs ~5-20x
# slower (frame_stack=4 trains at ~800 env-steps/s on this GPU). The timeout
# is therefore a generous fixed floor-rate ceiling, NOT derived from SPS_EST.
TRIAL_TIMEOUT_S = TRIAL_STEPS / 500 + 900


def objective(trial: optuna.Trial) -> float:
    """Train one short no-DR camera config and score its final eval.

    Args:
        trial: The Optuna trial supplying hyperparameters.

    Returns:
        ``mean_progress_m`` of the trial's final deterministic eval
        (NaN marks a crashed/timed-out trial as failed).
    """
    spec = base_spec_params(
        render=RENDER, num_envs=NUM_ENVS,
        lr=trial.suggest_float("lr", 3e-5, 6e-4, log=True),
        entropy_coef=trial.suggest_float("entropy_coef", 3e-4, 1e-2, log=True),
        epochs=trial.suggest_int("epochs", 3, 6),
        minibatches=trial.suggest_categorical("minibatches", [4, 8]),
        frame_stack=trial.suggest_categorical("frame_stack", [1, 4]),
        gamma=trial.suggest_categorical("gamma", [0.99, 0.995]),
        group="best_camera_hpo", variant=f"t{trial.number:03d}",
    )
    r = launch(f"trial_{trial.number:03d}", RUNNER_SRC,
               {"mode": "trial", "seed": SEED, "root": ROOT, "spec": spec},
               timeout_s=TRIAL_TIMEOUT_S)
    if "error" in r:
        print(f"  trial {trial.number}: FAILED ({r['error']})")
        return float("nan")
    m = r["metrics"]
    print(f"  trial {trial.number}: progress={m['mean_progress_m']:.2f}m "
          f"completion={m['completion_rate']:.2f} "
          f"({r['wall_s']/60:.1f} min)  {spec['lr']:.1e}/{spec['entropy_coef']:.1e}"
          f" ep{spec['epochs']} mb{spec['minibatches']} fs{spec['frame_stack']}")
    trial.set_user_attr("metrics", {k: v for k, v in m.items()
                                    if isinstance(v, (int, float))})
    return float(m["mean_progress_m"])


study = optuna.create_study(
    direction="maximize", study_name="best_camera",
    sampler=optuna.samplers.TPESampler(seed=SEED),
    storage=f"sqlite:///{ROOT}/study.db", load_if_exists=True)
if not study.trials:                       # anchor: sweep's best-known config
    study.enqueue_trial({"lr": 3e-4, "entropy_coef": 3e-3, "epochs": 5,
                         "minibatches": 8, "frame_stack": 4, "gamma": 0.99})

while len(study.trials) < HPO_TRIALS and elapsed_h() < HPO_DEADLINE_H:
    study.optimize(objective, n_trials=1)

print(f"\nHPO done: {len(study.trials)} trials in {elapsed_h():.2f}h; "
      f"best progress = {study.best_value:.2f} m")
print("best params:", study.best_params)
```

```python
"""Study results: every trial at a glance."""

import pandas as pd

rows = []
for t in study.trials:
    rows.append({"trial": t.number, "state": t.state.name,
                 "progress_m": t.value, **t.params})
trials_df = pd.DataFrame(rows).sort_values("progress_m", ascending=False)
display(trials_df.head(20))

BEST = dict(study.best_params)
print("winner:", BEST)
```

## Final training — winner config, track zoo, full DR

The winning hyperparameters now train long, with everything the physical
car needs: the demo zoo's 6 pre-baked reinvent variants (per-env world DR:
widths, palettes, fields, walls) split train/test with `reinvent_base` (the
printed track) held out of both, appearance/camera/physics/action DR, and
per-episode direction randomization. Telemetry, TensorBoard trajectory
figures, periodic deterministic evals, and the no-DR per-track holdout eval
all land automatically under the run dir.

The step budget is sized from *measured* throughput of the exact final spec
(a short bench of it below) against the wall-clock hours left, minus a
reserve for analysis and export.


```python
"""Size the final run from measured throughput, then train."""

from deepracer_genesis.datasets.splits import TrackDataset
from deepracer_genesis.tools.zoo import compile_zoo, demo_zoo

zoo_names = compile_zoo(demo_zoo())
split = TrackDataset(names=zoo_names, holdout=("reinvent_base",),
                     test_fraction=0.2, seed=SEED)
print(f"train={split.train}\ntest={split.test}\nholdout={split.holdout}")


def final_spec_params(steps: int, eval_every: int) -> dict:
    """The final run's spec params: winner HPO config + zoo + full DR.

    Args:
        steps: Total env-steps to train.
        eval_every: Periodic eval cadence in env-steps.

    Returns:
        Spec params for the payloads' ``build_spec``.
    """
    return base_spec_params(
        render=RENDER, num_envs=NUM_ENVS, tracks=list(split.train),
        dr=True, random_direction=True,
        real_tracks=list(split.test) + list(split.holdout),
        steps=steps, eval_every=eval_every,
        group="best_camera", variant="final", **BEST)


# measure the REAL final spec (zoo + DR is slower than the bare bench spec)
r = launch("bench_final", RUNNER_SRC, {
    "mode": "bench", "seed": SEED, "bench_steps": BENCH_STEPS,
    "compile_zoo": True,
    "spec": final_spec_params(steps=TRIAL_STEPS, eval_every=TRIAL_EVAL_EVERY),
}, timeout_s=1200)
sps_final = r.get("env_steps_per_s") or SPS_EST * 0.7
print(f"final-spec throughput: {sps_final:.0f} env-steps/s")

train_h = budget_left_h(reserve=FINAL_RESERVE_H)
FINAL_STEPS = int(min(FINAL_STEPS_MAX, train_h * 3600 * sps_final * 0.9))
FINAL_STEPS = max(FINAL_STEPS, FINAL_STEPS_MIN)
EVAL_EVERY = max(FINAL_STEPS // FINAL_EVALS, 500_000)
print(f"budget left {train_h:.2f}h -> FINAL_STEPS={FINAL_STEPS:,} "
      f"(eval every {EVAL_EVERY:,})")

final = launch("final", RUNNER_SRC, {
    "mode": "final", "seed": SEED, "root": ROOT, "compile_zoo": True,
    "spec": final_spec_params(steps=FINAL_STEPS, eval_every=EVAL_EVERY),
}, timeout_s=train_h * 3600 + 1800)
assert "error" not in final, f"final training failed: {final}"
RUN_DIR = final["run_dir"]
print(f"\nfinal run: {RUN_DIR}")
print("train:", final["train"])
print("metrics:", {k: round(v, 3) for k, v in final["metrics"].items()
                   if isinstance(v, (int, float))})
```

```python
"""Evaluation: learning curve, holdout table, trajectory plots."""

import pandas as pd

hist = pd.DataFrame(final["history"])
fig, ax = plt.subplots(1, 2, figsize=(12, 4))
ax[0].plot(hist["frames"], hist["mean_progress_m"], marker="o")
ax[0].set(xlabel="env steps", ylabel="mean progress (m)", title="eval progress")
ax[1].plot(hist["frames"], hist["completion_rate"], marker="o", color="tab:green")
ax[1].set(xlabel="env steps", ylabel="completion rate", title="eval completion")
fig.tight_layout(); plt.show()

if final.get("holdout"):
    display(pd.DataFrame(final["holdout"]).T[
        ["mean_progress_m", "completion_rate", "offtrack_rate", "mean_speed_mps"]])
```

```python
"""Trajectories from the recorded telemetry (final + holdout evals)."""

import io as _io

from IPython.display import Image as _Image, display as _display

from deepracer_genesis.analysis.telemetry import episode_agg, load_telemetry
from deepracer_genesis.analysis.trackplots import (plot_offtrack_hotspots,
                                                   plot_trajectories,
                                                   plot_variant_grid)


def show_fig(fig):
    """Render a figure inline regardless of the active matplotlib backend.

    Args:
        fig: The matplotlib figure to display (trackplots forces Agg, so
            plain ``plt.show()`` would render nothing in the notebook).
    """
    buf = _io.BytesIO()
    fig.savefig(buf, format="png", dpi=110, bbox_inches="tight")
    _display(_Image(buf.getvalue()))


df = load_telemetry(os.path.join(RUN_DIR, "telemetry", "final.parquet"))
display(episode_agg(df).sort_values("completion", ascending=False).head(12))
show_fig(plot_variant_grid(df, color_by="speed"))

hold_path = os.path.join(RUN_DIR, "telemetry", "holdout_reinvent_base.parquet")
if os.path.exists(hold_path):
    hdf = load_telemetry(hold_path)
    show_fig(plot_trajectories(hdf, "reinvent_base", color_by="speed"))
    show_fig(plot_offtrack_hotspots(hdf, "reinvent_base"))
```

```python
"""A rollout video of the trained policy on the printed track."""

vid = launch("video", RUNNER_SRC, {
    "mode": "video", "seed": SEED, "root": ROOT, "compile_zoo": True,
    "ckpt": os.path.join(RUN_DIR, "model.pt"), "video_steps": 400,
    "video_envs": 8,
    "spec": final_spec_params(steps=FINAL_STEPS, eval_every=EVAL_EVERY),
}, timeout_s=1800)
print(vid)
```

```python
"""ONNX export for the physical car (genesis-free subprocess; opset 11)."""

exp = launch("export", EXPORT_SRC, {
    "seed": SEED, "root": ROOT, "compile_zoo": True,
    "ckpt": os.path.join(RUN_DIR, "model.pt"), "bundle_name": "best_camera",
    "spec": final_spec_params(steps=FINAL_STEPS, eval_every=EVAL_EVERY),
}, timeout_s=900)
print(exp)
assert "error" not in exp, f"export failed: {exp}"
print(f"\ncar bundle: {exp['export_dir']}/best_camera.tar.gz")
print(f"total notebook wall-clock: {elapsed_h():.2f} h")
```

## Deployment & reproducibility

- **Car bundle**: `<run_dir>/export/best_camera.tar.gz` — `policy.onnx`
  (opset 11, OpenVINO-2021.1-compatible, `FRONT_FACING_CAMERA` input),
  `model_metadata.json`, and the model card. Upload to the car as usual.
- **TensorBoard**: `%tensorboard --logdir runs/best_camera` — scalar curves
  plus the `Eval/trajectories` and `Holdout/reinvent_base` figure timelines
  (Images tab) showing driving behavior evolve over training.
- **Reruns**: everything is seeded (`SEED` drives python/numpy/torch, the
  env, the TPE sampler, and the split); the Optuna study resumes from
  `runs/best_camera/study.db`, and every stage's exact params/log/result sit
  in `runs/best_camera/logs/`.
